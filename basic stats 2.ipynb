```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
### reading patient dataset
df=pd.read_csv('Patient_Data.csv')
```

```python
df.head()
```

```python
### patient dataset info()
df.info()
```

```python
### checking missing values
df.isnull().sum()
```

```python
#### fillling missing values with median 
#### missing value is float so filling median
df.fillna({'BillAmount':df.BillAmount.median()},inplace=True)
```

```python
df.isnull().sum()
```

```python
### checking duplicates
df.duplicated().sum()
```

```python
df[df.duplicated()]
```

```python
#### droping duplicates from dataset
df.drop_duplicates(inplace=True,ignore_index=True)
```

```python
df.duplicated().sum()
```

```python
#### droping ReceptionistID and checkInTime from dataset
df.drop(columns=['ReceptionistID','CheckInTime'],inplace=True)
```

```python
df.head()
```

```python
### calculating total bill department wise
total_bill_amount=df.groupby('Department')['BillAmount'].sum()
```

```python
total_bill_amount
```

```python
#### reading billing dataset
df1=pd.read_csv('Billing_Data.csv')
```

```python
df1
```

```python
##### merging billing dataset and patient dataset on patientID
merged_df=pd.merge(df,df1,on='PatientID')
```

```python
### after merging the dataset
merged_df
```

```python
### concatenating  additional dataframe that contains new patients details of current week 
new_patients=pd.DataFrame({
    'PatientID':[106,107],
    'Name':['John','Robin'],
    'Department':['Neurology','Cardiology'],
    'Doctor':['Dr.Smith','Dr.Lee'], 
    'BillAmount':[5400.0,6600.0]})
```

```python
#### new_patients dataset 
new_patients
```

```python
df.head()
```

```python
### concatenates new patient data with old patient data
df_updated=pd.concat([df,new_patients],ignore_index=True)
```

```python
#### after concatenates
df_updated
```

```python
##### concatenating an new billing category 
insurance=pd.DataFrame({
    'PatientID':[106,107],
    'InsuranceCovered':[2400.0,3000.0],
    'FinalAmount':[3000.0,3600.0]})
```

```python
df1_updated=pd.concat([df1,insurance],ignore_index=True)
```

```python
### after concatenates
df1_updated
```

```python
### final patient and billing dataset
final_updated_dataset=pd.merge(df_updated,df1_updated,on='PatientID')
```
